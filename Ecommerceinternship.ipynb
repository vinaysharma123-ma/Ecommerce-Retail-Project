# Handle Duplicate value

```python
print("Duplicates before:", df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()

# Reset index
df = df.reset_index(drop=True)

# Check again
print("Duplicates after:", df.duplicated().sum())
```

# Check Extra space b/w columns

```python
# Get text columns
text_columns = df.select_dtypes(include=["object"]).columns

# Remove leading/trailing spaces
for col in text_columns:
    df[col] = df[col].str.strip()

df.head()
```

# To remove multiple spaces between words:

```python
for col in text_columns:
    df[col] = df[col].str.replace(r"\s+", " ", regex=True)
df.head()
```

# For names/categories where Title Case is appropriate:

```python
for col in text_columns:
    df[col] = df[col].str.title()
df.head()
```

# Handle missing/null values

```python
# Missing values in each column
print(df.isnull().sum())

## Percentage of missing values:
missing_percentage = df.isnull().mean() * 100

print(missing_percentage)
```

# # Fill missing text values

```python
# For customer_name, we can use the most frequent value (mode):
# 
df["customer_name"] = df["customer_name"].fillna(
    df["customer_name"].mode()[0]
)
```

```python
## Check:

print(df["customer_name"].isnull().sum())
```

```python
# Fill missing numerical values
# For profit, use the median:
#
df["profit"] = df["profit"].fillna(
    df["profit"].median()
)
```

```python
#


#
Why median?

Median is less affected by extreme values than the mean, so it is often useful for financial columns such as profit.
```

```python
# Check:
print(df["profit"].isnull().sum())
```

# # Fix incorrect date formats

```python
#  dataset contains:

#order_date
# ship_date

# inseption:
print(df["order_date"].head())
print(df["ship_date"].head())
```

```python
# Convert them to proper datetime:
df["order_date"] = pd.to_datetime(
    df["order_date"],
    errors="coerce"
)

df["ship_date"] = pd.to_datetime(
    df["ship_date"],
    errors="coerce"
)
```

# Check invalid dates:

```python
print("Invalid order dates:",
      df["order_date"].isnull().sum())

print("Invalid ship dates:",
      df["ship_date"].isnull().sum())
```

```python
# Format dates consistently:
df["order_date"] = df["order_date"].dt.strftime("%Y-%m-%d")

df["ship_date"] = df["ship_date"].dt.strftime("%Y-%m-%d")
```

## Convert sales into numerical format

```python
df["sales"] = (
    df["sales"]
    .astype(str)
    .str.replace(",", "", regex=False)
)

df["sales"] = pd.to_numeric(
    df["sales"],
    errors="coerce"
)
```

```python
# Check the data type:
print(df["sales"].dtype)
```

```python
# Identify invalid numerical values
# Find negative profit values

#  profit should not contain invalid negative values if the business rule defines profit as non-negative.
# do not required for this code
negative_profit = df[df["profit"] < 0]

print(negative_profit)
```

```python
# Count them:
print("Negative profit records:",
      (df["profit"] < 0).sum())
```

# # Correct invalid negative profit

```python
# Correct invalid negative profit
# Using the median of valid profit values:
valid_profit_median = df.loc[
    df["profit"] >= 0,
    "profit"
].median()

df.loc[
    df["profit"] < 0,
    "profit"
] = valid_profit_median
```

```python
# Check again:
print(
    "Negative profit after cleaning:",
    (df["profit"] < 0).sum()
)
```

# Check for impossible values
# Check negative sales

```python
print(df[df["sales"] < 0])

# Check negative profit
print(df[df["profit"] < 0])

# Check zero/negative sales
print(df[df["sales"] <= 0])
```

```python
# For example, if sales must always be greater than zero:
invalid_sales = df[df["sales"] <= 0]

print(invalid_sales)
```

```python
# Check date consistency

# You can also check whether shipping occurred before ordering.
invalid_dates = df[
    df["ship_date"] < df["order_date"]
]

print(invalid_dates)
```

```python
# Complete Business Summary
```

```python

# Total Sales
total_sales = df['sales'].sum()

# Total Profit
total_profit = df['profit'].sum()

# Average Discount
avg_discount = df['discount'].mean()

# Highest revenue market
market_sales = df.groupby('market')['sales'].sum()
highest_market = market_sales.idxmax()
highest_market_revenue = market_sales.max()

# Least profitable category
category_profit_summary = df.groupby('category')['profit'].sum()
least_profitable_category = category_profit_summary.idxmin()
least_profitable_value = category_profit_summary.min()

# Most common shipping mode
ship_mode_counts = df['ship_mode'].value_counts()
most_used_ship_mode = ship_mode_counts.idxmax()
most_used_ship_count = ship_mode_counts.max()

# Highest discount category
category_discount_summary = df.groupby('category')['discount'].mean()
highest_discount_category = category_discount_summary.idxmax()
highest_discount_value = category_discount_summary.max()

# Best region
region_sales_summary = df.groupby('region')['sales'].sum()
best_region = region_sales_summary.idxmax()
best_region_sales = region_sales_summary.max()

print("=" * 60)
print("             BUSINESS ANALYSIS SUMMARY")
print("=" * 60)

print(f"Total Sales: {total_sales:,.2f}")
print(f"Total Profit: {total_profit:,.2f}")
print(f"Average Discount: {avg_discount:.2%}")

print("-" * 60)

print(
    f"Highest Revenue Market: {highest_market} "
    f"({highest_market_revenue:,.2f})"
)

print(
    f"Least Profitable Category: {least_profitable_category} "
    f"({least_profitable_value:,.2f})"
)

print(
    f"Most Common Shipping Mode: {most_used_ship_mode} "
    f"({most_used_ship_count:,} orders)"
)

print(
    f"Highest Discount Category: {highest_discount_category} "
    f"({highest_discount_value:.2%})"
)

print(
    f"Highest Sales Region: {best_region} "
    f"({best_region_sales:,.2f})"
)

print("=" * 60)
```
